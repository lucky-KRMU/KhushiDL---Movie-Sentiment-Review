## EDA (Exploratory Data Analysis) 🚀

```python
# importing the necessary modules
import tensorflow as tf
import pandas as pd
from pathlib import Path
```

```python
# path
data_path = Path("data/aclImdb")
test_path = data_path / "test"
train_path = data_path / "train"
```

```python
# data
pos_1 = test_path/'pos'
pos_2 = train_path/'pos'

neg_1 = test_path/'neg'
neg_2 = train_path/'neg'

pos = [x for x in pos_1.iterdir()] + [x for x in pos_2.iterdir()]
neg = [x for x in neg_1.iterdir()] + [x for x in neg_2.iterdir()]

print("Total Positives: {}".format(len(pos)))
print("Total Negatives: {}".format(len(neg)))
```
