## EDA (Exploratory Data Analysis) 🚀

```python
# importing the necessary modules
import tensorflow as tf
import pandas as pd
from pathlib import Path
```

```python
# path
data_path = Path("data/aclImdb")
test_path = data_path / "test"
train_path = data_path / "train"
```

```python
# data
pos_1 = test_path/'pos'
pos_2 = train_path/'pos'

neg_1 = test_path/'neg'
neg_2 = train_path/'neg'

pos = [x for x in pos_1.iterdir()] + [x for x in pos_2.iterdir()]
neg = [x for x in neg_1.iterdir()] + [x for x in neg_2.iterdir()]

print("Total Positives: {}".format(len(pos)))
print("Total Negatives: {}".format(len(neg)))
```

## Preprocessing(preparing) Data 📈

```python
# preparing the train data
train_path = Path('data/aclImdb/train')

train_data_raw = [x for x in (train_path).iterdir()]
# len(train_data_raw)

BATCH = 32
seed = 42

raw_train_ds = tf.keras.utils.text_dataset_from_directory(
    train_path,
    batch_size = BATCH,
    seed=seed
)
```

- NOTE: if there is a subset in text_dataset_from_directory then there must be a validation split

### Investigating the Classes (labels)

```python
for text_batch, label_batch in raw_train_ds.take(1):
    for i in range(5):
        print(f"Review: {text_batch.numpy()[i]}")
        print(f"Label: {label_batch.numpy()[i]}")
```

```python
# checking the labels
for i, label in enumerate(raw_train_ds.class_names):
    print(f"Label: {i} for Class Name : {label}")
```
